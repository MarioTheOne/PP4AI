````python
price = 350

if price > 500: # executes if condition is satisfied
    print("Premium")
elif price > 100: # executes if previous condition is not satisfied and this condition is satisfied
    print("Standard")
else: # executes if none of the previous conditions is satisfied
    print("Budget")
````

## **Loops in Python: Repeating actions**

Until now, every statement in our programs is written explicitly.

What if we want to perform the same operation on **every product** in our list?

````python
products = [
    {"name": "Laptop", "price": 1200, "stock": 5},
    {"name": "Mouse", "price": 25, "stock": 0},
    {"name": "Monitor", "price": 350, "stock": 12},
    {"name": "Keyboard", "price": 80, "stock": 7}
]

print(products[0]["name"])
print(products[1]["name"])
print(products[2]["name"])
print(products[3]["name"])
````

This works with four products.

But what if we had 100 products? Or 10,000?

We need a way to tell Python:

> **Repeat this operation for every element in a collection.**

### **The `for` loop**

````python
for product in products: # iterates over each element of the given collection
    print(product["name"]) # on each iteration executes action
````

During each iteration, the variable `product` temporarily refers to one element of the list.

Conceptually, Python does something like:

1. `product = products[0]`
2. Execute the indented block
3. `product = products[1]`
4. Execute the indented block
5. Continue until there are no elements left

#### What value has product during each iteration?

````python
i = 0 # iteration counter
for product in products:
    print(f"product at iteration {i}:", product)
    i += 1 # update the iteration counter before finishing the current iteration
````

## **The `while` loop**

A `for` loop is useful when we know what collection we want to iterate over or how many times we want to repeat something.

Sometimes instead we want to repeat an action **while a condition remains True**.

````python
attempts = 0

while attempts < 3:
    print("Attempt", attempts + 1)
    attempts = attempts + 1
````

attempts = 0 → condition True → execute

attempts = 1 → condition True → execute

attempts = 2 → condition True → execute

attempts = 3 → condition False → stop

Write a program that determines whether a word is a palindrome.

## **Returning values from functions**

So far, our functions have printed their results.

But printing a result and **returning** a result are not the same thing.

````python
def calculate_total(price, quantity):
    print(price * quantity)
````

````python
result = calculate_total(20, 3)

print("Result:", result)
````

What happened?

The function printed `60`, but the variable `result` contains `None`.

Why?

`print()` only displays information on the screen.

It does **not** send that value back to the part of the program that called the function.

If we want a function to produce a value that can be used later, we use `return`.

````python
def calculate_total(price, quantity):
    return price * quantity
````

````python
result = calculate_total(20, 3)

print("Result:", result)
````

Now the function **returns** the calculated value.

This means that we can store it:

````python
total = calculate_total(50, 4)

print(total)
````

Or use it directly in another expression:

````python
total_with_tax = calculate_total(50, 4) * 1.22

print(total_with_tax)
````

A useful way to think about functions is:

> **Inputs → Function → Output**

## **Functions can also contain loops**

````python
def factorial(n):
    result = 1

    for number in range(1, n + 1):
        result = result * number

    return result
````

2. The loop visits every product.
3. The condition checks whether the product is available.
4. The counter is updated when the condition is satisfied.
5. After the loop finishes, the function returns the final result.

## Recursion

````python
def factorial(n):
    # Base condition
    if n == 0:
        return 1
    else:
        # Call to the same function with a modified parameter
        return n * factorial(n - 1)
````

### **How does recursion work?**

A recursive function solves a problem by calling itself with a **smaller version of the same problem**.

Every recursive function needs two important parts:

1. A **base case** that stops the recursion.
2. A **recursive case** that calls the function again with a modified input.

For factorial:

```text
4! = 4 × 3!
3! = 3 × 2!
2! = 2 × 1!
1! = 1 × 0!
0! = 1
```

So calling:
```python
factorial(4)
```

Can be understood as:
```text
factorial(4)
→ 4 × factorial(3)
→ 4 × 3 × factorial(2)
→ 4 × 3 × 2 × factorial(1)
→ 4 × 3 × 2 × 1 × factorial(0)
→ 4 × 3 × 2 × 1 × 1
→ 24
```
